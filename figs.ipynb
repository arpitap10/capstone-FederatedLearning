```python
%cd /content/drive/MyDrive/capstone_project
import glob, pandas as pd, sys; sys.path.insert(0, "src")
import attacks
scores, coup = {}, {}
for f in glob.glob("runs/main__reliability__seed*/client_signals.csv"):
    seed = f.split("seed")[-1].split("/")[0]
    d = pd.read_csv(f)
    for c in d.client.unique():
        k = f"{c}__s{seed}"
        scores[k] = d[d.client == c].sort_values("round")["r"].values
        coup[k] = float(d[d.client == c]["coupling"].iloc[0])
print(attacks.attribute_inference_on_scores(scores, coup))
```
